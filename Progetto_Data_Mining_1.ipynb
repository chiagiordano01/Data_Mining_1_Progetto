#1.3 DATA QUALITY — Duplicati, Missing, Errori Semantici

```python
import pandas as pd
import numpy as np

# ============================================================
# LOAD DATASET
# ============================================================

df_raw = pd.read_csv("DM1_game_dataset.csv")
print("Totale righe iniziali:", len(df_raw))


# ============================================================
# 1. DUPLICATI
# ============================================================

print("\n--- DUPLICATI ---")
print("Duplicati su tutte le colonne:", df_raw.duplicated().sum())
print("Duplicati su BGGId:", df_raw["BGGId"].duplicated().sum())

df = df_raw.drop_duplicates(subset="BGGId", keep="first").copy()
print("Righe dopo rimozione duplicati:", len(df))


# ============================================================
# 2. MISSING ORIGINALI
# ============================================================

missing_original = df.isna().sum()

print("\n--- TOP 10 MISSING ORIGINALI ---")
print(missing_original.sort_values(ascending=False).head(10))


# ============================================================
# 3. ERRORI SEMANTICI
# ============================================================

print("\n--- REPORT ERRORI SEMANTICI ---")

# 3.1 ComWeight fuori [1,5]
mask_comweight = (df["ComWeight"] < 1) | (df["ComWeight"] > 5)
print("ComWeight fuori [1,5]:", mask_comweight.sum())
df.loc[mask_comweight, "ComWeight"] = np.nan


# 3.2 MinPlayers / MaxPlayers incoerenti
mask_players = (
    (df["MinPlayers"] <= 0) |
    (df["MaxPlayers"] <= 0) |
    (df["MaxPlayers"] < df["MinPlayers"])
)
print("MinPlayers / MaxPlayers incoerenti:", mask_players.sum())
df.loc[mask_players, ["MinPlayers", "MaxPlayers"]] = np.nan


# 3.3 YearPublished fuori range plausibile
mask_year = (df["YearPublished"] < 1850) | (df["YearPublished"] > 2025)
print("YearPublished fuori [1850,2025]:", mask_year.sum())
df.loc[mask_year, "YearPublished"] = np.nan


# 3.4 ComMinPlaytime / ComMaxPlaytime incoerenti
mask_playtime = (
    (df["ComMinPlaytime"] < 0) |
    (df["ComMaxPlaytime"] < 0) |
    (df["ComMaxPlaytime"] < df["ComMinPlaytime"])
)
print("ComMinPlaytime / ComMaxPlaytime incoerenti:", mask_playtime.sum())
df.loc[mask_playtime, ["ComMinPlaytime", "ComMaxPlaytime"]] = np.nan


# 3.5 Valori numerici impossibili (< 0)
numeric_cols = [
    "NumOwned", "NumWant", "NumWish",
    "NumUserRatings", "NumComments",
    "MfgPlayTime"
]

for col in numeric_cols:
    if col in df.columns:
        invalid = df[col] < 0
        print(f"{col} negativi:", invalid.sum())
        df.loc[invalid, col] = np.nan


# ============================================================
# 4. MISSING DOPO ERRORI SEMANTICI
# ============================================================

missing_after_errors = df.isna().sum()

print("\n--- TOP 10 MISSING DOPO ERRORI SEMANTICI ---")
print(missing_after_errors.sort_values(ascending=False).head(10))


# ============================================================
# DATA PREPARATION
# ============================================================

df_prep = df.copy()

print("\n--- DATA PREPARATION ---")

# 1. Imputazione con mediana (variabili numeriche/ordinali)
cols_impute_median = ["ComWeight", "ComAgeRec", "YearPublished"]

for col in cols_impute_median:
    if col in df_prep.columns:
        median_value = df_prep[col].median()
        df_prep[col] = df_prep[col].fillna(median_value)
        print(f"Imputazione mediana su {col}: {median_value}")


# 2. Rimozione righe con MinPlayers / MaxPlayers mancanti
rows_before = len(df_prep)
df_prep = df_prep.dropna(subset=["MinPlayers", "MaxPlayers"])
rows_after = len(df_prep)

print("Righe eliminate per MinPlayers/MaxPlayers mancanti:",
      rows_before - rows_after)

# 2. Rimozione righe con ComMinPlayime e ComMaxPlaytime
df_prep["ComMaxPlaytime"] = df_prep["ComMaxPlaytime"].fillna(
    df_prep["ComMaxPlaytime"].median()
)

# 3. LanguageEase: lasciata invariata (sarà esclusa dai modelli)
if "LanguageEase" in df_prep.columns:
    print("LanguageEase NON imputata (variabile esclusa dai modelli)")


print("\nShape finale dataset preparato:", df_prep.shape)
```

#2.1 K-MEANS

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import seaborn as sns

# ------------------------------------------------------------
# 1. Selezione variabili per il clustering
# ------------------------------------------------------------

cluster_features_raw = [
    "ComWeight",
    "ComMinPlaytime",
    "ComMaxPlaytime",
    "NumUserRatings",
    "NumWish"
]

# Rimuoviamo solo le righe con NaN nelle feature di clustering
df_cluster = df_prep.dropna(subset=cluster_features_raw).copy()

print("Righe totali df_prep:", df_prep.shape[0])
print("Righe usate per il clustering:", df_cluster.shape[0])

# ------------------------------------------------------------
# 2. Log-transform delle variabili skewed
# ------------------------------------------------------------

df_cluster["NumUserRatings_log"] = np.log10(df_cluster["NumUserRatings"] + 1)
df_cluster["NumWish_log"] = np.log10(df_cluster["NumWish"] + 1)

selected_features = [
    "ComWeight",
    "ComMinPlaytime",
    "ComMaxPlaytime",
    "NumUserRatings_log",
    "NumWish_log"
]

X = df_cluster[selected_features].copy()

print("Variabili usate per il clustering:")
print(selected_features)

# ------------------------------------------------------------
# 3. Normalizzazione Min-Max
# ------------------------------------------------------------

scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)


# ------------------------------------------------------------
# 3. Elbow Method (SSE)
# ------------------------------------------------------------

sse = []
K_range_elbow = range(2, 11)

for k in K_range_elbow:
    km = KMeans(
        n_clusters=k,
        init="k-means++",
        n_init=20,
        random_state=42
    )
    km.fit(X_scaled)
    sse.append(km.inertia_)

plt.figure(figsize=(7, 4))
plt.plot(K_range_elbow, sse, marker='o')
plt.title("Elbow Method – SSE")
plt.xlabel("k")
plt.ylabel("SSE")
plt.grid(True)
plt.show()

# ------------------------------------------------------------
# 4. Silhouette Score – solo nel range individuato dall’Elbow
# ------------------------------------------------------------

K_range_silhouette = range(3, 7)
silhouette_scores = []

for k in K_range_silhouette:
    km = KMeans(
        n_clusters=k,
        init="k-means++",
        n_init=20,
        random_state=42
    )
    labels = km.fit_predict(X_scaled)
    silhouette_scores.append(silhouette_score(X_scaled, labels))

plt.figure(figsize=(7, 4))
plt.plot(list(K_range_silhouette), silhouette_scores, marker='o')
plt.title("Silhouette Score (range basato su Elbow)")
plt.xlabel("k")
plt.ylabel("Silhouette")
plt.grid(True)
plt.show()

for k, s in zip(K_range_silhouette, silhouette_scores):
    print(f"k = {k} → silhouette = {s:.4f}")

# ---------------------------------
# 5. K-means finale con K ottimale
# ---------------------------------

BEST_K = 3

kmeans_final = KMeans(
    n_clusters=BEST_K,
    init="k-means++",
    n_init=20,
    random_state=42
)

cluster_labels = kmeans_final.fit_predict(X_scaled)
df_cluster["Cluster_KMeans"] = cluster_labels

# SSE e Silhouette finali
sse_final = kmeans_final.inertia_
silhouette_final = silhouette_score(X_scaled, cluster_labels)

print("\n=== RISULTATI FINALI K-MEANS ===")
print(f"K ottimale: {BEST_K}")
print(f"SSE finale: {sse_final:.4f}")
print(f"Silhouette finale: {silhouette_final:.4f}")

# ------------------------------------------------------------
# 6. Centroidi in unità originali
# ------------------------------------------------------------

centroids_scaled = kmeans_final.cluster_centers_
centroids_original = scaler.inverse_transform(centroids_scaled)

df_centroids = pd.DataFrame(centroids_original, columns=selected_features)
df_centroids.insert(0, "Cluster", [f"C{i}" for i in range(BEST_K)])

print("\nCentroidi (in unità originali):")
display(df_centroids)

# ------------------------------------------------------------
# 7. PCA – Visualizzazione dei cluster
# ------------------------------------------------------------

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
centroids_pca = pca.transform(kmeans_final.cluster_centers_)

plt.figure(figsize=(8, 6))
sns.scatterplot(
    x=X_pca[:, 0], y=X_pca[:, 1],
    hue=cluster_labels,
    palette="Set2",
    s=40, alpha=0.75
)

plt.scatter(
    centroids_pca[:, 0],
    centroids_pca[:, 1],
    c="black",
    s=250,
    marker="X",
    edgecolor="white",
    linewidth=2,
    label="Centroidi"
)

for i, (x, y) in enumerate(centroids_pca):
    plt.text(x + 0.015, y + 0.015, f"C{i}", fontsize=12, weight="bold")

plt.title(f"PCA – K-Means Clusters (k={BEST_K})")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.legend()
plt.grid(True)
plt.show()

# ------------------------------------------------------------
# 8. Parallel Coordinates – Centroidi
# ------------------------------------------------------------

from pandas.plotting import parallel_coordinates

plt.figure(figsize=(10, 6))
parallel_coordinates(
    df_centroids,
    class_column="Cluster",
    colormap="Set2",
    linewidth=2
)

plt.title("Parallel Coordinates – Profili dei centroidi (K-means)")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

#2.2 DBSCAN

```python
# =========================================
# STEP 1: Preprocessing per DBSCAN
# =========================================

import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# Stesse feature del K-means
cluster_features_raw = [
    "ComWeight",
    "ComMinPlaytime",
    "ComMaxPlaytime",
    "NumUserRatings",
    "NumWish"
]

# Drop delle righe con NaN SOLO per DBSCAN
df_db = df_prep.dropna(subset=cluster_features_raw).copy()

print("Righe totali df_prep:", df_prep.shape[0])
print("Righe usate per DBSCAN:", df_db.shape[0])

# Log-transform
df_db["NumUserRatings_log"] = np.log10(df_db["NumUserRatings"] + 1)
df_db["NumWish_log"] = np.log10(df_db["NumWish"] + 1)

features = [
    "ComWeight",
    "ComMinPlaytime",
    "ComMaxPlaytime",
    "NumUserRatings_log",
    "NumWish_log"
]

X = df_db[features].copy()

# Normalizzazione
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

```

```python
# ======================================
# STEP 2: K-distance plot (minPts = 4)
# ======================================

from sklearn.neighbors import NearestNeighbors
import matplotlib.pyplot as plt
import numpy as np

# minPts standard raccomandato
minPts = 4

# Modello per distanze
neighbors = NearestNeighbors(n_neighbors=minPts)
neighbors_fit = neighbors.fit(X_scaled)
distances, indices = neighbors_fit.kneighbors(X_scaled)

# Prendiamo per ogni punto la distanza al 4° vicino più prossimo
k_distances = np.sort(distances[:, -1])

# Plot
plt.figure(figsize=(7,4))
plt.plot(k_distances)
plt.title("K-distance Plot (minPts = 4)")
plt.xlabel("Punti ordinati")
plt.ylabel("Distanza al 4° vicino")
plt.grid(True)
plt.show()
```

```python
# =======================
# STEP 3: Clustering
# =======================

from sklearn.cluster import DBSCAN

# Valore scelto dal K-distance plot
eps_value = 0.13
minPts = 4

dbscan = DBSCAN(eps=eps_value, min_samples=minPts)
db_labels = dbscan.fit_predict(X_scaled)

df_db["DBSCAN_Cluster"] = db_labels

# Conteggio cluster
unique, counts = np.unique(db_labels, return_counts=True)

print("=== Risultati DBSCAN ===")
for u, c in zip(unique, counts):
    if u == -1:
        print(f"Noise (cluster -1): {c} punti")
    else:
        print(f"Cluster {u}: {c} punti")

print("\nNumero totale cluster (escluso noise):", len(unique) - (1 if -1 in unique else 0))
```

```python
# ========================================
# Test multipli con eps diversi
# ========================================

from sklearn.cluster import DBSCAN
import numpy as np

eps_list = [0.10, 0.08, 0.06]
minPts = 4

for eps_value in eps_list:

    db = DBSCAN(eps=eps_value, min_samples=minPts)
    labels = db.fit_predict(X_scaled)

    unique, counts = np.unique(labels, return_counts=True)

    print(f"\n===== eps = {eps_value} =====")
    for u, c in zip(unique, counts):
        if u == -1:
            print(f"Noise (cluster -1): {c} punti")
        else:
            print(f"Cluster {u}: {c} punti")

    print("Numero cluster (escluso noise):", len(unique) - (1 if -1 in unique else 0))
```

#2.3 HIERARCHICAL CLUSTERING

```python
# ===============
# DENDROGRAMMI
# ===============

from scipy.cluster.hierarchy import dendrogram, linkage
import matplotlib.pyplot as plt

methods = ["single", "ward", "complete", "average"]

plt.figure(figsize=(18, 5))

for i, method in enumerate(methods, 1):
    Z = linkage(X_scaled, method=method)

    plt.subplot(1, 4, i)
    dendrogram(
        Z,
        truncate_mode="lastp",
        p=20,
        leaf_rotation=90,
        leaf_font_size=10
    )
    plt.title(f"{method.capitalize()} linkage")
    plt.xlabel("Cluster (aggregati)")
    plt.ylabel("Distanza")

plt.tight_layout()
plt.show()
```

```python
# ==================================
# Estrazione cluster e silhouette
# ==================================

from scipy.cluster.hierarchy import fcluster
from sklearn.metrics import silhouette_score

Z_single = linkage(X_scaled, method="single")
Z_ward = linkage(X_scaled, method="ward")
Z_complete = linkage(X_scaled, method="complete")
Z_average = linkage(X_scaled, method="average")

k = 3

labels_single = fcluster(Z_single, k, criterion="maxclust")
labels_ward = fcluster(Z_ward, k, criterion="maxclust")
labels_complete = fcluster(Z_complete, k, criterion="maxclust")
labels_average = fcluster(Z_average, k, criterion="maxclust")

print("=== Silhouette Scores (k=3) ===")
print(f"Single   : {silhouette_score(X_scaled, labels_single):.4f}")
print(f"Ward     : {silhouette_score(X_scaled, labels_ward):.4f}")
print(f"Complete : {silhouette_score(X_scaled, labels_complete):.4f}")
print(f"Average  : {silhouette_score(X_scaled, labels_average):.4f}")
```

```python
# ============================================================
# TAGLIO DEL DENDROGRAMMA — AVERAGE LINKAGE (k = 3)
# ============================================================

plt.figure(figsize=(10, 6))

Z_avg = linkage(X_scaled, method="average")
dendrogram(Z_avg, truncate_mode="lastp", p=30)

cut_height = Z_avg[-(k-1), 2]
plt.axhline(y=cut_height, color="red", linestyle="--",
            label=f"Taglio (k={k})")

plt.title("Dendrogramma – Average linkage con taglio")
plt.xlabel("Cluster aggregati")
plt.ylabel("Distanza")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import seaborn as sns

# PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Plot clusters gerarchici (Average)
plt.figure(figsize=(8, 6))
sns.scatterplot(
    x=X_pca[:, 0],
    y=X_pca[:, 1],
    hue=labels_average,
    palette="Set2",
    s=40,
    alpha=0.75
)

plt.title("PCA – Hierarchical Clustering (Average Linkage, k=3)")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.grid(True)
plt.show()
```

```python
# Setup e feature engineering

import pandas as pd
import numpy as np

# Copia del dataset
df_clf = df_prep.copy()

# Log-transform (per ridurre skewness)
df_clf["NumWish_log"] = np.log10(df_clf["NumWish"] + 1)
df_clf["NumUserRatings_log"] = np.log10(df_clf["NumUserRatings"] + 1)

# Feature selezionate
features = [
    "ComWeight",
    "YearPublished",
    "MinPlayers",
    "MaxPlayers",
    "ComAgeRec",
    "NumUserRatings_log",
    "NumWish_log",
    "ComMinPlaytime",
    "ComMaxPlaytime"
]

target = "Rating"

X = df_clf[features]
y = df_clf[target]

print("Shape X:", X.shape)
print("\nDistribuzione target:")
print(y.value_counts())
print("\nMissing values (X):")
print(X.isna().sum())

print("\nMissing values dopo imputazione:")
print(X.isna().sum())
```

```python
# Encoding del target (Rating)

from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
y_enc = le.fit_transform(y)

print("Mapping classi:")
for cls, code in zip(le.classes_, le.transform(le.classes_)):
    print(f"{cls} → {code}")
```

```python
# Train / Test split

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_enc,
    test_size=0.25,
    random_state=42,
    stratify=y_enc
)

# Imputazione basata SOLO sul training set (evita leakage)
median_vals = X_train.median()
X_train = X_train.fillna(median_vals)
X_test = X_test.fillna(median_vals)

print("Train shape:", X_train.shape)
print("Test shape:", X_test.shape)

print("\nDistribuzione classi nel train:")
print(pd.Series(y_train).value_counts(normalize=True))

print("\nDistribuzione classi nel test:")
print(pd.Series(y_test).value_counts(normalize=True))
```

```python
# Valutazione del modello

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)
import seaborn as sns
import matplotlib.pyplot as plt

# Predizioni
y_pred = dt_baseline.predict(X_test)

# Accuracy
acc = accuracy_score(y_test, y_pred)
print(f"Accuracy (Decision Tree baseline): {acc:.4f}")

# Classification report
print("\nClassification Report:")
print(classification_report(y_test, y_pred, target_names=le.classes_))

# Confusion matrix
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(6, 5))
sns.heatmap(
    cm, annot=True, fmt="d",
    xticklabels=le.classes_,
    yticklabels=le.classes_,
    cmap="Blues"
)
plt.xlabel("Predicted")
plt.ylabel("True")
plt.title("Confusion Matrix – Decision Tree (baseline)")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np

# ============================================================
# LOAD DATASET
# ============================================================

df_raw = pd.read_csv("DM1_game_dataset.csv")
print("Totale righe iniziali:", len(df_raw))


# ============================================================
# 1. DUPLICATI
# ============================================================

print("\n--- DUPLICATI ---")
print("Duplicati su tutte le colonne:", df_raw.duplicated().sum())
print("Duplicati su BGGId:", df_raw["BGGId"].duplicated().sum())

df = df_raw.drop_duplicates(subset="BGGId", keep="first").copy()
print("Righe dopo rimozione duplicati:", len(df))


# ============================================================
# 2. MISSING ORIGINALI
# ============================================================

missing_original = df.isna().sum()

print("\n--- TOP 10 MISSING ORIGINALI ---")
print(missing_original.sort_values(ascending=False).head(10))


# ============================================================
# 3. ERRORI SEMANTICI
# ============================================================

print("\n--- REPORT ERRORI SEMANTICI ---")

# 3.1 ComWeight fuori [1,5]
mask_comweight = (df["ComWeight"] < 1) | (df["ComWeight"] > 5)
print("ComWeight fuori [1,5]:", mask_comweight.sum())
df.loc[mask_comweight, "ComWeight"] = np.nan


# 3.2 MinPlayers / MaxPlayers incoerenti
mask_players = (
    (df["MinPlayers"] <= 0) |
    (df["MaxPlayers"] <= 0) |
    (df["MaxPlayers"] < df["MinPlayers"])
)
print("MinPlayers / MaxPlayers incoerenti:", mask_players.sum())
df.loc[mask_players, ["MinPlayers", "MaxPlayers"]] = np.nan


# 3.3 YearPublished fuori range plausibile
mask_year = (df["YearPublished"] < 1850) | (df["YearPublished"] > 2025)
print("YearPublished fuori [1850,2025]:", mask_year.sum())
df.loc[mask_year, "YearPublished"] = np.nan


# 3.4 ComMinPlaytime / ComMaxPlaytime incoerenti
mask_playtime = (
    (df["ComMinPlaytime"] < 0) |
    (df["ComMaxPlaytime"] < 0) |
    (df["ComMaxPlaytime"] < df["ComMinPlaytime"])
)
print("ComMinPlaytime / ComMaxPlaytime incoerenti:", mask_playtime.sum())
df.loc[mask_playtime, ["ComMinPlaytime", "ComMaxPlaytime"]] = np.nan


# 3.5 Valori numerici impossibili (< 0)
numeric_cols = [
    "NumOwned", "NumWant", "NumWish",
    "NumUserRatings", "NumComments",
    "MfgPlayTime"
]

for col in numeric_cols:
    if col in df.columns:
        invalid = df[col] < 0
        print(f"{col} negativi:", invalid.sum())
        df.loc[invalid, col] = np.nan


# ============================================================
# 4. MISSING DOPO ERRORI SEMANTICI
# ============================================================

missing_after_errors = df.isna().sum()

print("\n--- TOP 10 MISSING DOPO ERRORI SEMANTICI ---")
print(missing_after_errors.sort_values(ascending=False).head(10))


# ============================================================
# DATA PREPARATION
# ============================================================

df_prep = df.copy()

print("\n--- DATA PREPARATION ---")

# 1. Imputazione con mediana (variabili numeriche/ordinali)
cols_impute_median = ["ComWeight", "ComAgeRec", "YearPublished"]

for col in cols_impute_median:
    if col in df_prep.columns:
        median_value = df_prep[col].median()
        df_prep[col] = df_prep[col].fillna(median_value)
        print(f"Imputazione mediana su {col}: {median_value}")


# 2. Rimozione righe con MinPlayers / MaxPlayers mancanti
rows_before = len(df_prep)
df_prep = df_prep.dropna(subset=["MinPlayers", "MaxPlayers"])
rows_after = len(df_prep)

print("Righe eliminate per MinPlayers/MaxPlayers mancanti:",
      rows_before - rows_after)


# 3. LanguageEase: lasciata invariata (sarà esclusa dai modelli)
if "LanguageEase" in df_prep.columns:
    print("LanguageEase NON imputata (variabile esclusa dai modelli)")


print("\nShape finale dataset preparato:", df_prep.shape)
```

```python
# --- 1. VARIABILI CONTINUE (qcut) ---
# 4 quartili come nel notebook (Light/MediumLight/MediumHeavy/Heavy ecc.)

labels_weight = ["Light", "MediumLight", "MediumHeavy", "Heavy"]
df["ComWeight_Bin"] = pd.qcut(df["ComWeight"], 4, labels=labels_weight).astype(str) + "_ComWeight"

labels_pop = ["Low", "Medium", "High", "VeryHigh"]
df["NumUserRatings_Bin"] = pd.qcut(df["NumUserRatings"], 4, labels=labels_pop).astype(str) + "_UserRatings"

df["NumWish_Bin"] = pd.qcut(df["NumWish"], 4, labels=labels_pop).astype(str) + "_NumWish"



# --- 2. CONTEGGI -> Has/No (come Expansions/Implementations/Alternates nel notebook) ---
df["Expansions_Bin"] = np.where(df["NumExpansions"] > 0, "HasExpansions", "NoExpansions")
df["Implementations_Bin"] = np.where(df["NumImplementations"] > 0, "HasImplementations", "NoImplementations")
df["Alternates_Bin"] = np.where(df["NumAlternates"] > 0, "HasAlternates", "NoAlternates")


# --- 3. PLAYTIME (cut manuale come nel notebook) ---
# uso ComMaxPlaytime come proxy della durata principale (puoi anche usare ComMin/ComMax separati)
bins_time = [0, 30, 90, 180, 70000]
labels_time = ["Rapid(<30)", "Standard(30-90)", "Long(90-180)", "VeryLong(>180)"]
df["ComMaxPlaytime_Bin"] = pd.cut(df["ComMaxPlaytime"], bins=bins_time, labels=labels_time).astype(str) + "_ComMaxTime"
df["ComMinPlaytime_Bin"] = pd.cut(df["ComMinPlaytime"], bins=bins_time, labels=labels_time).astype(str) + "_ComMinTime"


bins_year = [-5000, 1950, 1980, 1990, 2000, 2010, 2026]
labels_year = ["Ancient", "Classic", "80s", "90s", "00s", "10s+"]

df["Year_Bin"] = pd.cut(df["YearPublished"], bins=bins_year, labels=labels_year).astype(str) + "_Year"



# --- 5. AGE RECOMMENDATION (sia ComAgeRec che MfgAgeRec) ---
bins_age = [0, 7, 13, 99]
labels_age = ["Kids(<8)", "PreTeen(8-13)", "Teen/Adult(14+)"]
df["MfgAgeRec_Bin"] = pd.cut(df["MfgAgeRec"], bins=bins_age, labels=labels_age).astype(str) + "_MfgAge"
df["ComAgeRec_Bin"] = pd.cut(df["ComAgeRec"], bins=bins_age, labels=labels_age).astype(str) + "_ComAge"


# --- 6. LANGUAGE EASE + BEST PLAYERS (discrete -> bin “semplice” fedele allo spirito) ---
# LanguageEase spesso è una scala discreta: la trasformo in 3 fasce
bins_lang = [-1, 2, 4, 10]
labels_lang = ["Hard", "Medium", "Easy"]
df["LanguageEase_Bin"] = pd.cut(df["LanguageEase"], bins=bins_lang, labels=labels_lang).astype(str) + "_LangEase"

df["BestPlayers_Bin"] = df["BestPlayers"].astype("Int64").astype(str) + "_BestPlayers"

# --- 7. PLAYER RANGE (funzione come nel notebook, ma rinominata) ---
def discretize_players_range(min_p, max_p):
    if min_p == 1 and max_p == 1:
        return "Solo"
    elif min_p == 2 and max_p == 2:
        return "Duel"
    elif max_p <= 5:
        return "SmallGroup"
    else:
        return "Party"

df["PlayerRange_Bin"] = df.apply(lambda row: discretize_players_range(row["MinPlayers"], row["MaxPlayers"]), axis=1) + "_PlayerRange"


# --- 8. Boolean già in stringhe: aggiungo suffissi per uniformità (opzionale ma utile) ---
df["IsReimplementation_Bin"] = df["IsReimplementation"].astype(str) + "_Reimpl"
df["Kickstarted_Bin"] = df["Kickstarted"].astype(str) + "_Kickstarter"

df.head()
```

```python
from collections import Counter
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score, classification_report, confusion_matrix
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ==========================================
# 0) INPUT ATTESO
# ==========================================
# X_rules: lista di transazioni (liste di item) che include anche la classe Rating
# Esempio item dentro ogni record:
#   "..._Bin", "HasExpansions", ..., "High_Rating" / "Medium_Rating" / "Low_Rating"

target_classes = {"Low_Rating", "Medium_Rating", "High_Rating"}

# ==========================================
# 1) TRAIN/TEST SPLIT + DEFAULT CLASS
# ==========================================
x_train, x_test = train_test_split(X_rules, test_size=0.25, random_state=42)

train_labels = [next((item for item in rec if item in target_classes), None) for rec in x_train]
train_labels = [lbl for lbl in train_labels if lbl is not None]

default_class = Counter(train_labels).most_common(1)[0][0]
print(f"Default Class (majority in train): {default_class}")

# ==========================================
# 2) MINING RULES + PRUNING (ibrido lift/support)
# ==========================================
print("Esecuzione mining regole...")

# Soglie (puoi cambiarle in base a quante regole ottieni)
# - supp: in percentuale (fim)
# - conf: in percentuale
supp = 2      # prova 2, 5, 10 se troppo / troppo poco
conf = 30     # prova 20-60
zmin = 2      # regole non banali

rules_raw = apriori(x_train, target="r", supp=supp, zmin=zmin, conf=conf, report="aScl")

rules_df = pd.DataFrame(
    rules_raw,
    columns=["consequent", "antecedent", "abs_supp", "%_supp", "conf", "lift"]
)

# Teniamo solo regole che predicono il target e con lift > 1 (associazione positiva)
rules_df = rules_df[rules_df["consequent"].isin(target_classes) & (rules_df["lift"] > 1)]

print(f"Regole candidate (target + lift>1): {len(rules_df)}")

# Se non hai regole, abbassa conf o supp
if len(rules_df) == 0:
    print("ATTENZIONE: 0 regole dopo il filtro. Prova ad abbassare supp/conf oppure verificare che X_rules includa Rating.")
else:
    # Top per LIFT (qualità)
    top_lift = rules_df.sort_values(by="lift", ascending=False).groupby("consequent").head(50)

    # Top per SUPPORT (copertura)
    top_supp = rules_df.sort_values(by="%_supp", ascending=False).groupby("consequent").head(50)

    # Unione + deduplica
    combined_rules = pd.concat([top_lift, top_supp]).drop_duplicates(subset=["consequent", "antecedent"])
    print(f"Regole totali ibride (top lift + top supp): {len(combined_rules)}")

    rules_pruned = combined_rules

    # Preparo regole per voting
    voting_rules = []
    for _, row in rules_pruned.iterrows():
        voting_rules.append({
            "consequent": row["consequent"],
            "antecedent": set(row["antecedent"]),
            "lift": float(row["lift"]),
            "conf": float(row["conf"]),
        })

    print(f"Regole usate per il voto: {len(voting_rules)}")

# ==========================================
# 3) FUNZIONE DI PREDIZIONE (WEIGHTED VOTING)
# ==========================================
def predict_voting(record_features, rules, default_class):
    record_set = set(record_features)

    # punteggi per ogni classe
    score_board = {cls: 0.0 for cls in target_classes}
    rules_triggered = 0

    for rule in rules:
        if rule["antecedent"].issubset(record_set):
            score_board[rule["consequent"]] += rule["lift"]  # peso = lift
            rules_triggered += 1

    if rules_triggered == 0:
        return default_class

    return max(score_board, key=score_board.get)

# ==========================================
# 4) VALUTAZIONE SU TEST
# ==========================================
y_true = []
y_pred = []
uncovered_count = 0

print("Inizio predizione (Voting)...")

for record in x_test:
    true_label = next((item for item in record if item in target_classes), None)
    if true_label is None:
        continue  # record senza label (non dovrebbe succedere se X_rules è corretto)

    record_features = [item for item in record if item not in target_classes]

    pred = predict_voting(record_features, voting_rules, default_class)

    y_true.append(true_label)
    y_pred.append(pred)

    # copertura: almeno una regola scatta?
    matches_any = any(rule["antecedent"].issubset(set(record_features)) for rule in voting_rules)
    if not matches_any:
        uncovered_count += 1

# Metriche
print("-" * 40)
f1 = f1_score(y_true, y_pred, average="weighted")
coverage = ((len(y_true) - uncovered_count) / len(y_true)) * 100 if len(y_true) > 0 else 0

print(f"F1-Score (Weighted): {f1:.4f}")
print(f"Copertura Regole: {coverage:.2f}%")

print("\n--- Classification Report ---")
print(classification_report(y_true, y_pred))

cm = confusion_matrix(y_true, y_pred, labels=list(target_classes))
print("\n--- Confusion Matrix ---")
print(cm)

plt.figure(figsize=(6, 4))
sns.heatmap(cm, annot=True, fmt="d",
            xticklabels=list(target_classes),
            yticklabels=list(target_classes),
            cmap='Blues')
plt.title("Confusion Matrix (Rule-based Voting)")
plt.xlabel("Predicted")
plt.ylabel("True")
plt.tight_layout()
plt.show()
```

```python
from collections import Counter
from sklearn.metrics import f1_score, classification_report, confusion_matrix

# --- 1. PREPARAZIONE ---
# X deve essere una lista di liste (transazioni)
target_classes = {"High_Rating", "Medium_Rating", "Low_Rating"}

x_train, x_test = train_test_split(X_rules, test_size=0.25, random_state=0)

# Calcoliamo la Default Class (Majority Class)
train_labels = [next((item for item in rec if item in target_classes), None) for rec in x_train]
default_class = Counter(train_labels).most_common(1)[0][0]
print(f"Default Class: {default_class}")


# --- 2. MINING E PRUNING INTELLIGENTE ---
print("Esecuzione Apriori...")
# Usiamo zmin=2 per evitare regole vuote o banali
rules_raw = fpgrowth(x_train, target="r", supp=supp, zmin=2, conf=30, report="aScl")

rules_df = pd.DataFrame(
    rules_raw,
    columns=["consequent", "antecedent", "abs_supp", "%_supp", "conf", "lift"]
)

# Filtriamo solo le regole che predicono il target
rules_df = rules_df[rules_df["consequent"].isin(target_classes) & (rules_df["lift"] > 1)]

# 1. Calcoliamo le Top per LIFT (Qualità)
top_lift = rules_df.sort_values(by="lift", ascending=False).groupby("consequent").head(10)

# 2. Calcoliamo le Top per SUPPORT (Quantità/Copertura)
top_supp = rules_df.sort_values(by="%_supp", ascending=False).groupby("consequent").head(10)

# 3. Uniamo i due set e rimuoviamo duplicati
# (Usiamo l'ID della regola o resettiamo l'indice per unire)
combined_rules = pd.concat([top_lift, top_supp]).drop_duplicates(subset=["consequent", "antecedent"])

print(f"Regole totali ibride: {len(combined_rules)}")


rules_pruned = combined_rules

voting_rules = []
for _, row in rules_pruned.iterrows():
    voting_rules.append({
        "consequent": row["consequent"],
        "antecedent": set(row["antecedent"]),
        "lift": row["lift"],   # Il "peso" del voto
        "conf": row["conf"]    # Alternativa: usare confidenza
    })

print(f"Regole usate per il voto: {len(voting_rules)}")

# --- 2. FUNZIONE DI PREDIZIONE (WEIGHTED VOTING) ---

def predict_voting(record, rules, default_class):
    record_set = set(record)

    # Inizializziamo i punteggi per ogni classe a 0
    # Assicurati che score_board contenga tutte le tue classi
    score_board = {cls: 0 for cls in target_classes}

    rules_triggered = 0

    for rule in rules:
        # Se la regola si applica al record
        if rule["antecedent"].issubset(record_set):
            # Aggiungiamo il Lift al punteggio della classe predetta
            # (Puoi provare anche rule["conf"] o rule["lift"] * rule["conf"])
            score_board[rule["consequent"]] += rule["lift"]
            rules_triggered += 1

    # Se nessuna regola si è attivata, usiamo il default
    if rules_triggered == 0:
        return default_class

    # Restituiamo la classe con il punteggio più alto
    return max(score_board, key=score_board.get)

# --- 3. VALUTAZIONE ---
correct = 0
total = 0
uncovered_count = 0

# --- LISTE PER SALVARE I RISULTATI ---
y_true = []
y_pred = []

# Variabili per statistiche interne (opzionale)
uncovered_count = 0

print("Inizio predizione (Voting)...")

for record in x_test:
    # 1. Estrazione Ground Truth
    true_label = next((item for item in record if item in target_classes), None)

    # 2. Pulizia record per il test
    record_features = [item for item in record if item not in target_classes]

    if true_label:
        # Eseguiamo la predizione
        prediction = predict_voting(record_features, voting_rules, default_class)

        # Salviamo nelle liste per sklearn
        y_true.append(true_label)
        y_pred.append(prediction)

        # Controllo di copertura (se nessuna regola si è attivata)
        matches_any = any(rule["antecedent"].issubset(set(record_features)) for rule in voting_rules)
        if not matches_any:
            uncovered_count += 1

# --- CALCOLO METRICHE ---

# 1. F1 Score Globale (Weighted è meglio se le classi sono sbilanciate)
f1 = f1_score(y_true, y_pred, average = 'macro')
f1_w = f1_score(y_true, y_pred, average='weighted')

print("-" * 30)
print(f"F1-Score: {f1:.4f}")
print(f"F1-Score (Weighted): {f1_w:.4f}")
print(f"Copertura Regole: {((len(y_true) - uncovered_count) / len(y_true))*100:.2f}%")

# 2. Report Dettagliato (F1 per ogni classe)
print("\n--- Classification Report ---")
print(classification_report(y_true, y_pred))

# 3. Matrice di Confusione (Opzionale ma utilissima per capire gli errori)
# Ti mostra se il modello confonde classi vicine (es. High vs VeryHigh)
cm = confusion_matrix(y_true, y_pred, labels=list(target_classes))
print("\n--- Confusion Matrix ---")
#heatmap per visualizzare meglio
sns.heatmap(cm, annot=True, fmt='d', xticklabels=target_classes, yticklabels=target_classes, cmap='Blues')
plt.xlabel("Predicted Label", fontsize=14)
plt.ylabel("True Label", fontsize=14)
plt.title("Confusion Matrix", fontsize=16)
plt.show()
print(cm)
```
